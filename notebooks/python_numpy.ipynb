## 3.4 Functions

Functions allow us to package reusable logic into a single block of code.

Instead of repeatedly writing the same calculation, we can create a function and call it whenever required.

In this project, functions will be used to create reusable airline-level performance calculations.

```python
def calculate_delay_rate(data, delay_column="ARR_DELAY", threshold=30):
    delayed_flights = (data[delay_column] > threshold).sum()
    total_flights = data[delay_column].notna().sum()
    
    if total_flights == 0:
        return 0
    
    return (delayed_flights / total_flights) * 100
calculate_delay_rate(df)
```

```python
def airline_delay_rate(data, airline_name, threshold=30):
    airline_data = data[data["AIRLINE"] == airline_name]
    
    return calculate_delay_rate(
        airline_data,
        delay_column="ARR_DELAY",
        threshold=threshold
    )

airline_delay_rate(df, "Delta Air Lines Inc.")
```

## 3.5 Reusable Airline Performance Function

We will create a reusable function that calculates the main operational KPIs for any airline.

This reduces repeated code and makes the analysis easier to maintain.

```python
def calculate_airline_metrics(data, airline_name):
    airline_data = data[data["AIRLINE"] == airline_name]
    
    total_flights = len(airline_data)
    
    metrics = {
        "AIRLINE": airline_name,
        "TOTAL_FLIGHTS": total_flights,
        "AVERAGE_DEP_DELAY": airline_data["DEP_DELAY"].mean(),
        "AVERAGE_ARR_DELAY": airline_data["ARR_DELAY"].mean(),
        "CANCELLED_FLIGHTS": airline_data["CANCELLED"].sum(),
        "DIVERTED_FLIGHTS": airline_data["DIVERTED"].sum(),
        "ARR_DELAY_OVER_30_RATE": calculate_delay_rate(
            airline_data,
            "ARR_DELAY",
            30
        )
    }
    
    return metrics
    
calculate_airline_metrics(df, "Southwest Airlines Co.")
```

### Applying the Function to All Airlines

The function can now be reused for every airline in the dataset.

```python
airline_metrics_from_function = []

for airline in airlines:
    airline_metrics_from_function.append(
        calculate_airline_metrics(df, airline)
    )

airline_metrics_from_function
```

## 3.7 Object-Oriented Programming (OOP)

Object-Oriented Programming organizes data and related operations inside classes and objects.

For this project, we will create a simple AirlinePerformanceAnalyzer class.

The class will store the dataset and provide methods for airline-level analysis.

```python
class AirlinePerformanceAnalyzer:
    
    def __init__(self, data):
        self.data = data
    
    def airline_data(self, airline_name):
        return self.data[self.data["AIRLINE"] == airline_name]
    
    def average_arrival_delay(self, airline_name):
        airline_data = self.airline_data(airline_name)
        return airline_data["ARR_DELAY"].mean()
    
    def cancellation_rate(self, airline_name):
        airline_data = self.airline_data(airline_name)
        
        total_flights = len(airline_data)
        
        if total_flights == 0:
            return 0
        
        return (airline_data["CANCELLED"].sum() / total_flights) * 100
```

```python
analyzer = AirlinePerformanceAnalyzer(df)

analyzer
```

```python
analyzer.average_arrival_delay("Delta Air Lines Inc.")
```

```python
analyzer.cancellation_rate("Delta Air Lines Inc.")
```

## 3.8 Exception Handling

Exception handling prevents the program from failing unexpectedly when an error occurs.

In data analytics, exception handling is useful when:
- a required column is missing
- a file cannot be loaded
- invalid input is supplied
- a calculation cannot be performed

```python
def safe_airline_delay(data, airline_name):
    try:
        if "AIRLINE" not in data.columns:
            raise KeyError("AIRLINE column is missing")
        
        if "ARR_DELAY" not in data.columns:
            raise KeyError("ARR_DELAY column is missing")
        
        airline_data = data[data["AIRLINE"] == airline_name]
        
        if airline_data.empty:
            return None
        
        return airline_data["ARR_DELAY"].mean()
    
    except KeyError as error:
        return str(error)
```

```python
safe_airline_delay(df, "Delta Air Lines Inc.")
```

```python
safe_airline_delay(df, "Unknown Airline")
```

## 4.10 Rule-Based Delay Severity Classification

We will use `np.where()` to classify flights based on arrival delay.

Rules:

- Delay <= 15 minutes → ON_TIME
- Delay > 15 and <= 30 → MODERATE_DELAY
- Delay > 30 and <= 60 → SIGNIFICANT_DELAY
- Delay > 60 → SEVERE_DELAY

This is a transparent business-rule algorithm, not machine learning.

```python
delay_severity = np.where(
    arrival_delay_array <= 15,
    "ON_TIME",
    np.where(
        arrival_delay_array <= 30,
        "MODERATE_DELAY",
        np.where(
            arrival_delay_array <= 60,
            "SIGNIFICANT_DELAY",
            "SEVERE_DELAY"
        )
    )
)

delay_severity[:20]
```

```python
severity_labels, severity_counts = np.unique(
    delay_severity,
    return_counts=True
)

severity_summary = dict(
    zip(severity_labels, severity_counts)
)

severity_summary
```
